```python

# ============================================================
# CEE SAATHI AI — MEDICAL ENTRANCE INTELLIGENCE SYSTEM
# Version 1.0 | Educational analytics prototype
# Python + Pandas + NumPy + Seaborn + Plotly + Scikit-learn
# ============================================================

from pathlib import Path
from datetime import date
import json
import sqlite3
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display, Markdown
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, classification_report, roc_auc_score
)
from sklearn.inspection import permutation_importance

warnings.filterwarnings("ignore")
np.random.seed(42)

# -------------------- CONFIGURATION --------------------

PROJECT = Path("CEE-Saathi-AI")
DATA = PROJECT / "data"
EXPORTS = PROJECT / "exports"

DATA.mkdir(parents=True, exist_ok=True)
EXPORTS.mkdir(parents=True, exist_ok=True)

N_STUDENTS = 1500
EXAM_DATE = date(2027, 6, 1)  # DEMO DATE: replace with verified exam date
PASS_THRESHOLD = 50            # DEMO THRESHOLD: not official CEE policy
MAX_MARKS = 100
DB_PATH = PROJECT / "cee_saathi.db"

display(Markdown("""
# 🎓 CEE SAATHI AI
### Medical Entrance Preparation Intelligence Dashboard

**Mode:** Educational demo and scenario analysis

This notebook uses synthetic student data. Its pass/fail labels, predictions,
rank estimates, scores and thresholds are not official CEE results.
"""))

# -------------------- 1. GENERATE DEMO DATA --------------------

def generate_demo_data(n=1500, seed=42):
    rng = np.random.default_rng(seed)

    study_hours = np.clip(rng.normal(4.5, 2.0, n), 0, 12)
    preparation_days = rng.integers(7, 240, n)
    mock_attempts = rng.integers(0, 40, n)
    attendance = np.clip(rng.normal(78, 15, n), 20, 100)
    topic_completion = np.clip(rng.normal(65, 22, n), 0, 100)
    revision_count = rng.integers(0, 15, n)

    preparation_signal = (
        study_hours * 3.2
        + np.log1p(preparation_days) * 3
        + mock_attempts * 0.45
        + attendance * 0.12
        + topic_completion * 0.22
        + revision_count * 0.65
    )

    base = 8 + preparation_signal
    noise = rng.normal(0, 11, n)

    physics = np.clip(base * 0.29 + noise + rng.normal(0, 8, n), 0, 100)
    chemistry = np.clip(base * 0.31 + noise + rng.normal(0, 8, n), 0, 100)
    biology = np.clip(base * 0.35 + noise + rng.normal(0, 7, n), 0, 100)

    total_score = np.clip(
        (physics + chemistry + biology) / 3
        + rng.normal(0, 5, n),
        0, MAX_MARKS
    )

    previous_score = np.clip(total_score - rng.normal(3, 10, n), 0, 100)
    latest_score = total_score
    score_trend = latest_score - previous_score

    # Labels are synthetic teaching labels, not real exam outcomes.
    demo_pass = (total_score >= PASS_THRESHOLD).astype(int)

    df = pd.DataFrame({
        "student_id": [f"DEMO-{i:05d}" for i in range(1, n + 1)],
        "is_demo_data": True,
        "study_hours_per_day": study_hours.round(2),
        "preparation_days": preparation_days,
        "mock_tests_attempted": mock_attempts,
        "attendance_percentage": attendance.round(2),
        "topic_completion_percentage": topic_completion.round(2),
        "revision_count": revision_count,
        "physics_score": physics.round(2),
        "chemistry_score": chemistry.round(2),
        "biology_score": biology.round(2),
        "previous_mock_score": previous_score.round(2),
        "latest_mock_score": latest_score.round(2),
        "score_trend": score_trend.round(2),
        "total_score": total_score.round(2),
        "demo_outcome": demo_pass
    })

    return df


students = generate_demo_data(N_STUDENTS)

students.to_csv(DATA / "demo_students.csv", index=False)

display(Markdown("## 1. Dataset preview"))
display(students.head(10))
print("Number of demo students:", len(students))
print("Dataset saved:", (DATA / "demo_students.csv").resolve())

# -------------------- 2. DATA QUALITY CHECK --------------------

def data_quality_report(df):
    return pd.DataFrame({
        "column": df.columns,
        "dtype": [str(x) for x in df.dtypes],
        "missing_values": df.isna().sum().values,
        "unique_values": df.nunique().values
    })

display(Markdown("## 2. Data quality report"))
display(data_quality_report(students))

assert students["student_id"].is_unique
assert students["total_score"].between(0, MAX_MARKS).all()
assert students["demo_outcome"].isin([0, 1]).all()

print("Basic data validation passed.")

# -------------------- 3. EXECUTIVE DASHBOARD --------------------

total_students = len(students)
passed = int((students["total_score"] >= PASS_THRESHOLD).sum())
failed = total_students - passed
pass_rate = passed / total_students * 100
avg_score = students["total_score"].mean()
highest_score = students["total_score"].max()
avg_study = students["study_hours_per_day"].mean()

display(Markdown("## 3. Executive summary"))

summary = pd.DataFrame({
    "Metric": [
        "Total demo students",
        "Pass scenario count",
        "Fail scenario count",
        "Pass scenario rate",
        "Average score",
        "Highest score",
        "Average daily study hours"
    ],
    "Value": [
        total_students,
        passed,
        failed,
        f"{pass_rate:.2f}%",
        f"{avg_score:.2f}",
        f"{highest_score:.2f}",
        f"{avg_study:.2f}"
    ]
})

display(summary)

# -------------------- 4. SUBJECT ANALYTICS --------------------

subject_columns = [
    "physics_score",
    "chemistry_score",
    "biology_score"
]

subject_summary = pd.DataFrame({
    "Subject": ["Physics", "Chemistry", "Biology"],
    "Average": [students[c].mean() for c in subject_columns],
    "Median": [students[c].median() for c in subject_columns],
    "Minimum": [students[c].min() for c in subject_columns],
    "Maximum": [students[c].max() for c in subject_columns],
    "Std Dev": [students[c].std() for c in subject_columns]
}).round(2)

display(Markdown("## 4. Subject performance"))
display(subject_summary)

# -------------------- 5. PROFESSIONAL STATIC CHARTS --------------------

sns.set_theme(style="whitegrid", context="notebook")

fig, axes = plt.subplots(2, 2, figsize=(16, 10))

sns.histplot(
    students["total_score"], bins=25, kde=True,
    ax=axes[0, 0]
)
axes[0, 0].axvline(
    PASS_THRESHOLD, linestyle="--",
    label=f"Demo threshold: {PASS_THRESHOLD}"
)
axes[0, 0].set_title("Score Distribution")
axes[0, 0].legend()

sns.barplot(
    data=subject_summary,
    x="Subject", y="Average",
    ax=axes[0, 1]
)
axes[0, 1].set_title("Average Score by Subject")

sns.countplot(
    x=np.where(
        students["total_score"] >= PASS_THRESHOLD,
        "Pass scenario", "Fail scenario"
    ),
    ax=axes[1, 0]
)
axes[1, 0].set_title("Pass/Fail Scenario Counts")

sns.scatterplot(
    data=students.sample(min(500, len(students)), random_state=42),
    x="study_hours_per_day",
    y="total_score",
    hue="demo_outcome",
    ax=axes[1, 1]
)
axes[1, 1].set_title("Study Hours vs Demo Score")

plt.tight_layout()
plt.show()

# -------------------- 6. CORRELATION ANALYSIS --------------------

display(Markdown("## 6. Correlation analysis"))

numeric = students.select_dtypes(include=np.number)
correlation = numeric.corr()

plt.figure(figsize=(13, 9))
sns.heatmap(
    correlation,
    cmap="coolwarm",
    center=0,
    annot=False
)
plt.title("Feature Correlation Heatmap")
plt.tight_layout()
plt.show()

display(
    correlation[["total_score"]]
    .sort_values("total_score", ascending=False)
)

# Correlation does not establish causation.

# -------------------- 7. PASS-THRESHOLD SCENARIOS --------------------

display(Markdown("## 7. Hypothetical threshold sensitivity"))

thresholds = [30, 40, 50, 60, 70, 80]
threshold_rows = []

for threshold in thresholds:
    count = int((students["total_score"] >= threshold).sum())
    threshold_rows.append({
        "hypothetical_threshold": threshold,
        "students_above_threshold": count,
        "students_below_threshold": total_students - count,
        "percentage_above": round(count / total_students * 100, 2)
    })

threshold_df = pd.DataFrame(threshold_rows)
display(threshold_df)

plt.figure(figsize=(10, 5))
sns.lineplot(
    data=threshold_df,
    x="hypothetical_threshold",
    y="percentage_above",
    marker="o"
)
plt.title("How the Hypothetical Threshold Changes the Cohort")
plt.xlabel("Hypothetical score threshold")
plt.ylabel("Percentage above threshold")
plt.tight_layout()
plt.show()

# -------------------- 8. MACHINE LEARNING --------------------

display(Markdown("""
## 8. Machine-learning demonstration

The target label is generated from a synthetic score threshold.
The following models demonstrate a workflow; they are not validated
predictors of real CEE admission or real student success.
"""))

feature_columns = [
    "study_hours_per_day",
    "preparation_days",
    "mock_tests_attempted",
    "attendance_percentage",
    "topic_completion_percentage",
    "revision_count",
    "physics_score",
    "chemistry_score",
    "biology_score",
    "previous_mock_score",
    "score_trend"
]

X = students[feature_columns].copy()
y = students["demo_outcome"].copy()

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

models = {
    "Dummy baseline": DummyClassifier(strategy="most_frequent"),
    "Logistic Regression": Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
        ("model", LogisticRegression(max_iter=2000))
    ]),
    "Random Forest": Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("model", RandomForestClassifier(
            n_estimators=250,
            min_samples_leaf=3,
            random_state=42,
            class_weight="balanced"
        ))
    ])
}

model_results = []
fitted_models = {}

for name, model in models.items():
    model.fit(X_train, y_train)
    predictions = model.predict(X_test)

    row = {
        "Model": name,
        "Accuracy": accuracy_score(y_test, predictions),
        "Precision": precision_score(y_test, predictions, zero_division=0),
        "Recall": recall_score(y_test, predictions, zero_division=0),
        "F1": f1_score(y_test, predictions, zero_division=0)
    }

    if hasattr(model, "predict_proba") and len(np.unique(y_test)) == 2:
        probabilities = model.predict_proba(X_test)[:, 1]
        row["ROC_AUC"] = roc_auc_score(y_test, probabilities)
    else:
        row["ROC_AUC"] = np.nan

    model_results.append(row)
    fitted_models[name] = model

model_comparison = pd.DataFrame(model_results).sort_values(
    "F1", ascending=False
)

display(model_comparison.round(4))

model_comparison.to_csv(
    EXPORTS / "model_comparison_demo.csv",
    index=False
)

# -------------------- 9. CONFUSION MATRIX --------------------

best_model_name = model_comparison.iloc[0]["Model"]
best_model = fitted_models[best_model_name]
best_predictions = best_model.predict(X_test)

cm = confusion_matrix(y_test, best_predictions)

plt.figure(figsize=(6, 5))
sns.heatmap(
    cm, annot=True, fmt="d",
    cmap="Blues"
)
plt.title(f"Confusion Matrix — {best_model_name}")
plt.xlabel("Predicted label")
plt.ylabel("Actual synthetic label")
plt.tight_layout()
plt.show()

print("Best demo model:", best_model_name)
print(classification_report(
    y_test,
    best_predictions,
    zero_division=0
))

# -------------------- 10. PERMUTATION IMPORTANCE --------------------

display(Markdown("## 10. Feature importance"))

try:
    importance = permutation_importance(
        best_model,
        X_test,
        y_test,
        n_repeats=5,
        random_state=42,
        scoring="f1"
    )

    importance_df = pd.DataFrame({
        "Feature": feature_columns,
        "Importance": importance.importances_mean
    }).sort_values("Importance", ascending=False)

    display(importance_df)

    plt.figure(figsize=(10, 6))
    sns.barplot(
        data=importance_df,
        x="Importance",
        y="Feature"
    )
    plt.title("Permutation Importance on Synthetic Test Data")
    plt.tight_layout()
    plt.show()

except Exception as error:
    print("Feature importance unavailable:", error)

# -------------------- 11. HYPOTHETICAL RANK PERCENTILE --------------------

display(Markdown("""
## 11. Demo-cohort rank scenario

This is a position within the generated demo cohort only.
It is not a national CEE rank or an admission prediction.
"""))

def demo_rank_scenario(score, cohort_scores):
    scores = pd.Series(cohort_scores).dropna()
    if scores.empty:
        raise ValueError("Cohort scores are empty.")

    percentile = float((scores <= score).mean() * 100)
    position = int((scores > score).sum() + 1)

    return {
        "input_score": float(score),
        "demo_cohort_percentile": round(percentile, 2),
        "approx_demo_position": position,
        "demo_cohort_size": int(len(scores))
    }

example_score = 65
display(pd.DataFrame([
    demo_rank_scenario(example_score, students["total_score"])
]))

# -------------------- 12. PERSONALIZED STUDY PLAN --------------------

display(Markdown("## 12. Personalized revision planner"))

def create_study_plan(
    subject_accuracies,
    available_hours=4.0,
    days_remaining=30
):
    if available_hours <= 0:
        raise ValueError("Available hours must be positive.")
    if days_remaining < 0:
        raise ValueError("Days remaining cannot be negative.")

    rows = []

    for subject, accuracy in subject_accuracies.items():
        accuracy = float(accuracy)
        if not 0 <= accuracy <= 100:
            raise ValueError("Accuracy must be between 0 and 100.")

        if accuracy < 50:
            priority = "VERY HIGH"
            weight = 4
            action = "Revise concepts and solve guided MCQs"
        elif accuracy < 70:
            priority = "HIGH"
            weight = 3
            action = "Analyze mistakes and practise targeted MCQs"
        elif accuracy < 85:
            priority = "MEDIUM"
            weight = 2
            action = "Mixed practice and spaced revision"
        else:
            priority = "MAINTAIN"
            weight = 1
            action = "Timed practice and short revision"

        rows.append({
            "Subject": subject,
            "Accuracy": accuracy,
            "Priority": priority,
            "Suggested Action": action,
            "Weight": weight
        })

    plan = pd.DataFrame(rows)

    if plan.empty:
        return plan

    plan["Hours Per Day"] = (
        plan["Weight"] / plan["Weight"].sum() * available_hours
    ).round(2)

    plan["Days Remaining"] = days_remaining

    return plan.sort_values(
        ["Weight", "Accuracy"],
        ascending=[False, True]
    )

demo_accuracy = {
    "Physics": float(students["physics_score"].mean()),
    "Chemistry": float(students["chemistry_score"].mean()),
    "Biology": float(students["biology_score"].mean())
}

study_plan = create_study_plan(
    demo_accuracy,
    available_hours=5,
    days_remaining=max((EXAM_DATE - date.today()).days, 0)
)

display(study_plan)
study_plan.to_csv(EXPORTS / "study_plan_demo.csv", index=False)

# -------------------- 13. SEAT SCENARIO CALCULATOR --------------------

display(Markdown("""
## 13. Hypothetical admission-seat scenario

Enter verified figures when available. The default values below are
fictional examples, not current Nepal medical admission seat counts.
"""))

def calculate_seat_scenario(total_seats, hypothetical_rank):
    total_seats = int(total_seats)
    hypothetical_rank = int(hypothetical_rank)

    if total_seats < 0:
        raise ValueError("Total seats cannot be negative.")
    if hypothetical_rank < 1:
        raise ValueError("Rank must be at least 1.")

    return {
        "hypothetical_total_seats": total_seats,
        "hypothetical_rank": hypothetical_rank,
        "rank_within_seat_count": hypothetical_rank <= total_seats,
        "warning": "Scenario only; not an admission guarantee."
    }

display(pd.DataFrame([
    calculate_seat_scenario(total_seats=100, hypothetical_rank=80)
]))

# -------------------- 14. SQLITE STORAGE --------------------

display(Markdown("## 14. Local SQLite storage"))

with sqlite3.connect(DB_PATH) as conn:
    students.to_sql(
        "demo_students",
        conn,
        if_exists="replace",
        index=False
    )
    model_comparison.to_sql(
        "model_comparison",
        conn,
        if_exists="replace",
        index=False
    )
    study_plan.to_sql(
        "study_plan",
        conn,
        if_exists="replace",
        index=False
    )

print("SQLite database saved:", DB_PATH.resolve())

# -------------------- 15. EXPORT REPORTS --------------------

report = {
    "project": "CEE Saathi AI",
    "version": "1.0",
    "generated_on": date.today().isoformat(),
    "data_mode": "SYNTHETIC DEMO DATA",
    "number_of_demo_students": int(total_students),
    "hypothetical_threshold": int(PASS_THRESHOLD),
    "pass_scenario_count": int(passed),
    "fail_scenario_count": int(failed),
    "pass_scenario_percentage": round(pass_rate, 2),
    "average_score": round(float(avg_score), 2),
    "highest_score": round(float(highest_score), 2),
    "best_demo_model": best_model_name,
    "important_warning": (
        "Synthetic demonstration only. Not validated for real CEE prediction, "
        "official rank, seat allocation or admission outcomes."
    )
}

with open(EXPORTS / "project_report.json", "w", encoding="utf-8") as file:
    json.dump(report, file, indent=2, ensure_ascii=False)

students.to_csv(EXPORTS / "students_export_demo.csv", index=False)
subject_summary.to_csv(EXPORTS / "subject_summary_demo.csv", index=False)
threshold_df.to_csv(EXPORTS / "threshold_sensitivity_demo.csv", index=False)

display(Markdown("## 15. Final project report"))
display(pd.DataFrame([report]).T.rename(columns={0: "Value"}))

print("\n" + "=" * 60)
print("CEE SAATHI AI — DEMO RUN FINISHED")
print("=" * 60)
print("Students:", total_students)
print("Pass scenario:", passed)
print("Fail scenario:", failed)
print("Average score:", round(avg_score, 2))
print("Best demo model:", best_model_name)
print("Output folder:", EXPORTS.resolve())
print("\nREMINDER: all generated student records and model results are fictional.")
```
